```python
def f(x, y=2):
    return x ** y

print(f(3))
print(f(3, 3))
```

```python
def sum_all(*args):
    return sum(args)

print(sum_all(1, 2, 3))
print(sum_all(10, 20, 30, 40))
```

```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

print_details(name="Tanmay", age=20, city="Pune")
```

```python
def add(a, b):
    return a + b

print(add(5, 3)) # argument assigned by their position in the function call
```

```python
def greet(name, age):
    print(name, age)

greet(age=20, name="Tanmay") # argument are passed by using parmater names 
```

```python
def greet(name, message="Hello"):
    print(message, name)

greet("Tanmay")
greet("Gaurav", "Welcome") # parameter has a predefined value used when no argument is supplied for it
```

```python
def student(name, *, age):
    print(name, age)

student("Tanmay", age=20) # These parameters must be passed using their names. They are defined after * or after *args 
```

 # Recursion

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5)) # the base case is factorial(0) = 1 (also factorial(1) = 1). It stops the recursion.
```

```python
def fibonacci(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6)) # we use the convention that the sequence starts with 0, 1, so fibonacci(0) = 0 and fibonacci(1) = 1.
```

 # Scope and closures 

```python
def outer(x):
    def inner(y):
        return x + y
    return inner

f = outer(10)
print(f(5))
```

```python
 # A closure is an inner function that remembers variables from its enclosing 
 # function's scope, even after the enclosing function has finished executing.

def outer(x):
    def inner(y):
        return x + y
    return inner

add_ten = outer(10)

print(add_ten(5))
print(add_ten(20))
```

```python
x = 10

def outer():
    x = 20

    def inner():
        nonlocal x
        x += 5
    inner()
    return x

print(outer())
```
