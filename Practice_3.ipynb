```python
def gen():
    yield 1
    yield 2
    return 3

g = gen()

print(next(g))
print(next(g))

try:
    print(next(g))
except StopIteration as e:
    print('Done', e.value)
```

```python
def my_gen():
    yield 10
    yield 20

def my_func():
    return 10

print(list(my_gen()))
print(my_func())
```

```python
def even_numbers(limit):
    for i in range(0, limit + 1, 2):
        yield i

for num in even_numbers(10):
    print(num)
```

```python
def numbers():
    for i in range(1, 1000001):
        yield i

g = numbers()

print(next(g))
print(next(g))
print(next(g))
```

```python
def gen():
    try:
        yield 'A'
        yield 'B'
    finally:
        print('Cleanup')

g = gen()

print(next(g))
g.close()
```
